```python
import scanpy as sc
import anndata as ad
import numpy as np
import os, sys

DATA_PATH = os.path.join(os.getenv("MLAB"), "projects/brcameta/projects/sig_recon/data")
STACK_PATH = os.path.join(os.getenv("MLAB"), "projects/brcameta/projects/sig_recon/scripts/stack")
```

```python
k562_ctrl_data = sc.read(os.path.join(DATA_PATH, "stack/sciplex/k562_ctrl.h5ad"))
a549_ctrl_data = sc.read(os.path.join(DATA_PATH, "stack/sciplex/a549_ctrl.h5ad"))
mcf7_ctrl_data = sc.read(os.path.join(DATA_PATH, "stack/sciplex/mcf7_ctrl.h5ad"))

k562_all_data = sc.read(os.path.join(DATA_PATH, "stack/sciplex/k562_all.h5ad"))
a549_all_data = sc.read(os.path.join(DATA_PATH, "stack/sciplex/a549_all.h5ad"))
mcf7_all_data = sc.read(os.path.join(DATA_PATH, "stack/sciplex/mcf7_all.h5ad"))
```

```python
def update_var_index(adata, new='feature_name', old='ensembl_id'):
    adata.var[old] = adata.var.index.copy()
    adata.var.index = adata.var[new].astype(str)
    adata.var_names_make_unique()
    adata.var.drop(new, axis=1, inplace=True)
    return adata
```

```python
k562_ctrl_data = update_var_index(k562_ctrl_data)
a549_ctrl_data = update_var_index(a549_ctrl_data)
mcf7_ctrl_data = update_var_index(mcf7_ctrl_data)

k562_ctrl_data.obs["cell_line"] = "k562"
a549_ctrl_data.obs["cell_line"] = "a549"
mcf7_ctrl_data.obs["cell_line"] = "mcf7"

k562_all_data = update_var_index(k562_all_data)
a549_all_data = update_var_index(a549_all_data)
mcf7_all_data = update_var_index(mcf7_all_data)
```

```python
a549_k562_ctrl = ad.concat([a549_ctrl_data, k562_ctrl_data], axis=0, join="inner")
a549_mcf7_ctrl = ad.concat([a549_ctrl_data, mcf7_ctrl_data], axis=0, join="inner")
mcf7_k562_ctrl = ad.concat([mcf7_ctrl_data, k562_ctrl_data], axis=0, join="inner")
```

```python
k562_all_data.write(os.path.join(DATA_PATH, "stack/sciplex/k562_all.h5ad"))
a549_all_data.write(os.path.join(DATA_PATH, "stack/sciplex/a549_all.h5ad"))
mcf7_all_data.write(os.path.join(DATA_PATH, "stack/sciplex/mcf7_all.h5ad"))
k562_ctrl_data.write(os.path.join(DATA_PATH, "stack/sciplex/k562_ctrl.h5ad"))
a549_ctrl_data.write(os.path.join(DATA_PATH, "stack/sciplex/a549_ctrl.h5ad"))
mcf7_ctrl_data.write(os.path.join(DATA_PATH, "stack/sciplex/mcf7_ctrl.h5ad"))
```

```python
a549_k562_ctrl.write(os.path.join(DATA_PATH, "stack/sciplex/a549_k562_ctrl.h5ad"))
a549_mcf7_ctrl.write(os.path.join(DATA_PATH, "stack/sciplex/a549_mcf7_ctrl.h5ad"))
mcf7_k562_ctrl.write(os.path.join(DATA_PATH, "stack/sciplex/mcf7_k562_ctrl.h5ad"))
```
